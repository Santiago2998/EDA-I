```python
#Poner materias como diccionario
def crear_materia(nombre, inicio, fin, prioridad):
    materia = {"nombre": nombre, "inicio": inicio, "fin": fin, "prioridad": prioridad}
    return materia


# Ordeno las materias por la hora en que acaba usando insertion sort
def insertion_sort_por_fin(materias):
    for i in range(1, len(materias)):
        actual = materias[i]
        j = i - 1
        while j >= 0 and materias[j]["fin"] > actual["fin"]:
            materias[j + 1] = materias[j]
            j -= 1
        materias[j + 1] = actual
    return materias

# Checar si se empalman materias
def hay_traslape(materia_A, materia_B):
    if materia_A["inicio"] < materia_B["fin"] and materia_B["inicio"] < materia_A["fin"]:
        return True
    return False


# Greedy: siempre escojo primero la materia que termina mas temprano
# Incremental: voy agregando de una en una, verificando que no haya conflicto
def construir_horario(materias):
    if len(materias) == 0:
        return []
    
    materias_ordenadas = insertion_sort_por_fin(materias)

    print("\n Materias Ordenadas:")
    for m in materias_ordenadas:
        print(f"  {m['nombre']} termina a las {m['fin']}")

    horario = []
    horario.append(materias_ordenadas[0])

    for i in range(1, len(materias_ordenadas)):
        candidata = materias_ordenadas[i]
        ultima = horario[-1]

        if not hay_traslape(ultima, candidata):
            horario.append(candidata)
        else:
            # CASO 1: mismo horario exacto
            mismo_horario = (ultima["inicio"] == candidata["inicio"] and 
                             ultima["fin"] == candidata["fin"])

            if mismo_horario:
                if ultima["prioridad"] > candidata["prioridad"]:
                    continue  # se queda la actual
                elif ultima["prioridad"] < candidata["prioridad"]:
                    horario[-1] = candidata
                else:
                    # MISMA PRIORIDAD → preguntar
                    print("\nConflicto detectado (misma hora y prioridad):")
                    print(f"1. {ultima['nombre']}")
                    print(f"2. {candidata['nombre']}")

                    while True:
                        eleccion = input("¿Cuál prefieres? (1/2): ")
                        if eleccion == "1":
                            break
                        elif eleccion == "2":
                            horario[-1] = candidata
                            break
                        else:
                            print("Opción inválida.")

            else:
                # CASO 2: se empalman pero no son idénticas
                if candidata["prioridad"] > ultima["prioridad"]:
                    horario[-1] = candidata
                # si no, se mantiene la actual

    return horario


# Muestro el horario final
def mostrar_horario(horario):
    print("\n---HORARIO SELECCIONADO---")
    if len(horario) == 0:
        print("No se pudo armar ningun horario.")
        return

    print(f"{'Materia':<25} {'Inicio':>8} {'Fin':>8} {'Prioridad':>12}")
    print("\n")
    for m in horario:
        print(f"{m['nombre']:<25} {m['inicio']:>8} {m['fin']:>8} {m['prioridad']:>12}")
    print("\n")
    print(f"Total materias aceptadas: {len(horario)}")

def pedir_materias():
    materias = []
    print("=== Construccion de Horario Academico ===")
    print("Escribe 'listo' como nombre cuando termines de ingresar materias.\n")

    while True:
        nombre = input("Nombre de la materia: ").strip()
        if nombre.lower() == "listo":
            break
        if nombre == "":
            print("El nombre no puede estar vacio.")
            continue

        while True:
            try:
                inicio = float(input(f"Hora de inicio de '{nombre}' (7.5 para 7:30): "))
                break
            except ValueError:
                print("Escribe un numero valido.")

        while True:
            try:
                fin = float(input(f"Hora de fin de '{nombre}' (9.0 para 9:00): "))
                if fin <= inicio:
                    print("La hora de fin debe ser mayor que la de inicio.")
                else:
                    break
            except ValueError:
                print("Escribe un numero valido.")

        while True:
            try:
                prioridad = int(input(f"Prioridad de '{nombre}' (1 = baja, 5 = alta): "))
                if 1 <= prioridad <= 5:
                    break
                else:
                    print("La prioridad debe ser entre 1 y 5.")
            except ValueError:
                print("Escribe un numero entero.")

        materias.append(crear_materia(nombre, inicio, fin, prioridad))
        print(f"'{nombre}' registrada.\n")

    return materias

materias = pedir_materias()

horario = construir_horario(materias)
mostrar_horario(horario)
```
