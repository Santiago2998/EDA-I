# Marco Teórico

Python es un lenguaje de programación de alto nivel, interpretado y de
propósito general, diseñado para ser fácil de leer y escribir gracias a
su sintaxis clara y estructurada. Es ampliamente utilizado en diferentes
áreas como el desarrollo de software, análisis de datos, automatización
y aplicaciones científicas [1]. Una de sus principales características
es que es dinámicamente tipado, lo que significa que no es necesario
declarar el tipo de dato de una variable, ya que este se determina
automáticamente en tiempo de ejecución [2].

En el desarrollo de programas, las variables permiten almacenar
información en memoria y pueden contener distintos tipos de datos como
enteros, números decimales, cadenas de texto y valores booleanos. El
correcto manejo de estos tipos es fundamental para evitar errores y
garantizar el funcionamiento adecuado del programa [2]. Asimismo, Python
permite realizar operaciones aritméticas básicas como suma, resta,
multiplicación, división y potenciación, las cuales son esenciales para
resolver problemas matemáticos y modelar situaciones reales [1].

Para lograr la interacción con el usuario, se emplean mecanismos de
entrada y salida de datos, los cuales permiten recibir información y
mostrar resultados. Este proceso es fundamental en la construcción de
programas dinámicos e interactivos [2]. Además, Python cuenta con
bibliotecas especializadas que amplían sus capacidades, como las
funciones matemáticas que incluyen operaciones trigonométricas tales
como seno, coseno y tangente, ampliamente utilizadas en áreas
científicas e ingenieriles [3].

Otra funcionalidad importante es el manejo de archivos, el cual permite
crear, leer y modificar información almacenada de manera permanente en
dispositivos de almacenamiento. Esto resulta esencial para aplicaciones
que requieren persistencia de datos [1]. De igual manera, las cadenas de
texto son estructuras que permiten almacenar información textual y
pueden manipularse mediante diversas operaciones como reemplazo de
palabras, concatenación y manejo de texto multilínea, facilitando el
procesamiento de información [2]. También es posible extraer partes
específicas de una cadena mediante técnicas como el slicing, lo que
permite trabajar con fragmentos de texto de manera eficiente [1].

El formato de texto mediante f-strings representa una forma moderna y
eficiente de insertar valores dentro de cadenas, mejorando la
legibilidad del código y simplificando la manipulación de datos [2]. Por
otro lado, las estructuras de control permiten dirigir el flujo de
ejecución de un programa; las condicionales permiten tomar decisiones
basadas en condiciones, mientras que los ciclos permiten repetir
acciones múltiples veces, siendo fundamentales en la lógica de
programación [1].

Las estructuras de datos permiten organizar la información de manera
eficiente; por ejemplo, las pilas operan bajo el principio LIFO, las
colas bajo el principio FIFO y los diccionarios almacenan datos en pares
clave-valor, facilitando el acceso rápido a la información [2].
Asimismo, las funciones son bloques de código reutilizables que permiten
estructurar programas de manera modular; estas pueden ser definidas por
el usuario o ser funciones anónimas (lambda), útiles para operaciones
simples y rápidas [2]. Finalmente, la implementación de menús en
programas permite organizar la interacción con el usuario mediante
diferentes opciones, combinando estructuras de control, ciclos y entrada
de datos.

código en línea, bloques de código, citas, líneas horizontales, enlaces

## *Ejercicio #6 - Fragmentos de carácteres*
- Definiremos una variable que contenga una frase utilizando tripletas de comillas.
- A la misma variable se le establecerá un rango de carácteres que pueda imprimir **[a : b]**.
- Utilizando f-string imprimiremos estas cadenas limitadas por rangos, de esta  manera solo se imprimirán los carácteres encontrados en las posiciones que hemos especificado.

```python
frase = """La frase para tomar fragmentos de char"""

parte1 = frase[0:10]
parte2 = frase[10:19]
parte3 = frase[19:38]

print(f"La Frase completa es: {frase}")
print(f"La parte 1 de la frase es: {parte1}")
print("La parte 2 de la frasse es: {}".format(parte2))
print(f"La parte 3 de la frase es: {parte3}")
```

## *Ejercicio #7  - Elaborar un menú de opciones*
- Al utilizar las funciones **if** junto a **elif** realizaremos un menú primitivo el cual nos permitirá realizar las mismas operaciones de ejercicio 2.
- Con la opción para calcular potencias utilizaremos un ciclo, el cual multiplicará la base por sí misma la cantidad de veces que sea necesario para llegar al valor deseado.

```python

print("Menu Calculadora")
print("1. Suma\n2. Resta\n3. Multiplicación\n4. División\n5. Potencia")
opcion = int(input("Selecciona una opción: "))

if (opcion == 1):
    n1 = float(input("Ingresa un valor 1: "))
    n2 = float(input("Ingresa un valor 2: "))
    suma = n1 + n1
    print(f"La suma es: {suma}")
elif (opcion == 2):
    n1 = float(input("Ingresa un valor 1: "))
    n2 = float(input("Ingresa un valor 2: "))
    resta = n1 - n2
    print(f"La resta es: {resta}")
elif (opcion == 3):
    n1 = float(input("Ingresa un valor 1: "))
    n2 = float(input("Ingresa un valor 2: "))
    multip = n1 * n2
    print(f"La multiplicación es: {multip}")
elif (opcion == 4):
    n1 = float(input("Ingresa un valor 1: "))
    n2 = float(input("Ingresa un valor 2: "))
    division = n1 / n2
    print(f"La division es: {division}")
elif (opcion == 5):
    respuesta = int(input("Cuantas veces quieres hacer la potencia? "))
    contador = 0
    while (contador != respuesta):
        n1 = float(input("Ingresa la base: "))
        n2 = float(input("Ingresa el exponente: "))
        potencia = n1 ** n2
        print(f"La potencia es: {potencia}")
        contador += 1
else:
    print("No existe esa opción")
```

### *8.1 - Implementación de una Pila*
- La pila se define de la siguiente manera: **Pila = [  ]**.
- Con el subfijo **.append** se añadirán los elementos que decida el usuario.
- Utilizaremos el subfijo **.pop()**, de esta forma se eliminará el nodo que se encuentra en el tope de nuestra pila

```python
pila = []

pila.append("Dato 1")
pila.append("Dato 2")
pila.append("Dato 3")

ultimo = pila.pop() 

print(f"Salió: {ultimo}") 
print(f"Pila actual: {pila}")
```

### *8.2 - Implementación de una Queue simple*
- Importamos desde collections la biblioteca deque.
- Con el subfijo **.append** -para nuestra variable que defina a nuestra queue definida por la función **deque()**- se añadirán los elementos que decida el usuario.
- Asignaremos una variable que contenga el primer elemento de nuestra queue, para ello utilizaremos el subfijo **.popleft()**, de esta forma se eliminará el último valor almacenado de nuestra queue
- Se imprimirán el valor eliminado junto a los valores que permanecieron en nuestra queue.

```python
from collections import deque

cola = deque()

cola.append("Persona A")
cola.append("Persona B")
print(f"La cola simple es: {cola}")

primero = cola.popleft() 

print(f"Atendiendo a: {primero}")
print(f"Quedan en fila: {list(cola)}")
```

## *Ejercicio #9 - Función lambda*
- La función de tipo lambda suelen definirse en una sola línea, por lo que el código que ejecutan suele ser pequeño.
- Con el mismo menú del ejercicio anterior utilizaremos una función de tipo lambda para la multiplicación de nuestras variables.
- Nuestra función quedará definida de la siguiente manera:  
*multiplicar* = lambda a, b: a * b
- Para llamar a nuestra función solo debemos de hacer lo siguiente:  
**resultado = multiplicar(var1, var2)**

```python
# 1. Función declarada normal (con validación de seguridad)
def division_segura(a, b):
    if b == 0:
        return "Error: No se puede dividir entre cero"
    return a / b

# 2. Función Lambda (anónima y rápida)
multiplicar = lambda a, b: a * b

print("Menu Calculadora")
print("1. Suma\n2. Resta\n3. Multiplicación\n4. División\n5. Potencia")
opcion = int(input("Selecciona una opción: "))

# Operaciones comunes para evitar repetir código
if 1 <= opcion <= 4:
    n1 = float(input("Ingresa un valor 1: "))
    n2 = float(input("Ingresa un valor 2: "))

if opcion == 1:
    suma = n1 + n2
    print(f"La suma es: {suma}")

elif opcion == 2:
    resta = n1 - n2
    print(f"La resta es: {resta}")

elif opcion == 3:
    # Uso de la función Lambda
    resultado = multiplicar(n1, n2)
    print(f"La multiplicación es: {resultado}")

elif opcion == 4:
    # Uso de la función normal
    resultado = division_segura(n1, n2)
    print(f"La división es: {resultado}")

elif opcion == 5:
    respuesta = int(input("¿Cuántas veces quieres hacer la potencia? "))
    contador = 0
    while contador != respuesta:
        base = float(input("Ingresa la base: "))
        exponente = float(input("Ingresa el exponente: "))
        print(f"La potencia es: {base ** exponente}")
        contador += 1
else:
    print("No existe esa opción")
```
