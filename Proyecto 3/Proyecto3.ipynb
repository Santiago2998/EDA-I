# Marco Teórico

En la programación existen diferentes maneras de resolver problemas de una forma más organizada y eficiente. Para lograrlo se utilizan algoritmos, que básicamente son una serie de pasos ordenados que ayudan a encontrar una solución a un problema específico. En este proyecto se trabaja con el problema del cambio de monedas, el cual consiste en devolver una cantidad de dinero usando la menor cantidad posible de monedas disponibles. Aunque parece algo sencillo, en programación este problema ayuda a entender cómo diferentes estrategias pueden trabajar juntas para resolver una situación de manera rápida y eficiente. Además, este tipo de ejercicios son importantes porque permiten aplicar conceptos fundamentales de estructuras de datos y algoritmos que después se utilizan en programas más complejos [1].
Uno de los conceptos principales que se utiliza en este proyecto es el algoritmo Merge Sort, el cual sirve para ordenar datos. Este algoritmo pertenece a una técnica conocida como “divide y vencerás”, que consiste en dividir un problema grande en partes más pequeñas, resolverlas por separado y después unirlas nuevamente hasta obtener el resultado final [2]. En el caso de este proyecto, el ordenamiento es importante porque primero se deben acomodar las monedas de mayor a menor para que el programa pueda tomar primero las de mayor valor. Esto hace que el proceso sea más rápido y organizado, ya que evita estar revisando monedas al azar [3]. El funcionamiento de Merge Sort consiste en dividir una lista en partes más pequeñas hasta que solo quede un elemento, para después comenzar a compararlas y unirlas nuevamente en el orden correcto. Aunque existen otros algoritmos de ordenamiento, este se considera eficiente porque funciona bien incluso cuando se trabaja con una gran cantidad de datos [2].
Después del ordenamiento se utiliza una estrategia llamada greedy o voraz. Esta estrategia consiste en elegir la mejor opción disponible en ese momento con la intención de llegar a una buena solución final [4]. En otras palabras, el algoritmo toma decisiones rápidas sin revisar todas las posibilidades posibles. En el problema del cambio de monedas esto significa que el programa siempre escogerá primero la moneda de mayor valor que no se pase del monto restante. Por ejemplo, si se necesita devolver una cantidad de 87 y las monedas disponibles son de 50, 20, 10, 5, 2 y 1, el algoritmo primero tomaría una moneda de 50, luego una de 20, después una de 10, una de 5 y finalmente una de 2, logrando completar el monto con pocas monedas [5].  
Sin embargo, aunque la estrategia greedy suele ser rápida y fácil de implementar, no siempre garantiza la mejor solución en todos los casos. Existen sistemas de monedas donde elegir siempre la moneda más grande puede terminar usando más monedas de las necesarias. Por eso es importante aclarar que este método funciona correctamente cuando las denominaciones siguen ciertos patrones, como el sistema de monedas de [1, 5, 10, 20, 50], que es el esquema usado en este proyecto [5].  
Otro tema importante dentro de este proyecto es la recursividad. La recursividad es una técnica de programación donde una función puede llamarse a sí misma para resolver un problema paso a paso [6]. Esto puede sonar complicado al principio, pero realmente se trata de resolver un problema grande usando versiones más pequeñas del mismo problema hasta llegar a un punto donde ya no sea necesario seguir repitiendo el proceso. En este caso, la recursividad se utiliza cuando el programa selecciona una moneda y después vuelve a ejecutarse con el monto restante, repitiendo el procedimiento hasta que el valor llegue a cero o ya no sea posible seguir utilizando monedas [7]. Gracias a esto, el código se vuelve más ordenado y evita repetir instrucciones innecesarias, haciendo más fácil entender el funcionamiento del programa.  
También es importante hablar sobre la complejidad temporal y espacial de los algoritmos, ya que esto ayuda a entender qué tan eficiente es un programa [8]. La complejidad temporal se refiere al tiempo que tarda un algoritmo en ejecutarse dependiendo de la cantidad de datos que tenga que procesar, mientras que la complejidad espacial tiene que ver con la cantidad de memoria que necesita utilizar. En el caso de Merge Sort, se considera un algoritmo eficiente porque tiene una complejidad temporal de O(n log n), lo que significa que puede ordenar datos de manera relativamente rápida incluso cuando hay muchos elementos [9]. En cuanto a la estrategia greedy, normalmente tiene un tiempo de ejecución menor porque solo recorre las monedas disponibles y selecciona las más convenientes. La recursividad, por otro lado, utiliza memoria extra debido a las llamadas que se van almacenando temporalmente mientras el problema se sigue resolviendo.
En general, este proyecto permite entender cómo diferentes estrategias de programación pueden combinarse para resolver un problema real de manera eficiente. Mientras Merge Sort organiza las monedas disponibles, la estrategia greedy ayuda a seleccionar las monedas más convenientes y la recursividad permite repetir el proceso automáticamente hasta completar el cambio necesario. Todo esto ayuda a desarrollar una mejor comprensión sobre cómo funcionan los algoritmos y por qué es importante elegir métodos eficientes al momento de programar.

# Desarrollo

En esta práctica se realizó un programa que simule un sistema de cambio.

## *Evaluación - Sistema de cambio*
- Implementar la estrategia divide y vencerás.
- Implementar la estrategia greedy para elegir la moneda más conveniente en cada paso.
- Resolver la construcción del cambio mediante recursividad.
- Analizar la complejidad en tiempo y en espacio tanto de cada estrategia como de la solución completa.

```python
import time
import tracemalloc
import matplotlib.pyplot as plt

def merge_sort(st):     # O(n log n)    S(n)
    if len(st) <= 1:
        return st
    mid = len(st) // 2

    izquierda = merge_sort(st[:mid])
    derecha = merge_sort(st[mid:])

    return merge(izquierda, derecha)

def merge(izq, der):
    resultado = []
    i = j = 0

    while i < len(izq) and j < len(der):
        if izq[i] > der[j]:
            resultado.append(izq[i])
            i += 1
        else:
            resultado.append(der[j])
            j += 1

    resultado.extend(izq[i:])
    resultado.extend(der[j:])
    
    return resultado

def cambio_recursivo(monto, monedas, indice):   # O(n)  S(n)
    if monto == 0:
        return []
    if indice >= len(monedas):
        return []
    moneda_actual = monedas[indice]
    if moneda_actual <= monto:
        return [moneda_actual] + cambio_recursivo(monto - moneda_actual, monedas, indice)
    else:
        return cambio_recursivo(monto, monedas, indice + 1)

def analizar_complejidad():
    tamanos = [10, 100, 500, 1000, 2000]
    tiempos = []
    memorias = []

    for n in tamanos:
        datos = list(range(n, 0, -1)) 

        # medir tiempo
        inicio = time.time()
        merge_sort(datos)
        fin = time.time()
        tiempos.append(fin - inicio)

        # medir memoria
        tracemalloc.start()
        merge_sort(datos)
        memoria_actual, memoria_pico = tracemalloc.get_traced_memory()
        tracemalloc.stop()
        memorias.append(memoria_pico / 1024)

    # Graficar tiempo
    plt.figure(figsize=(10,5))
    plt.plot(tamanos, tiempos, marker='o', label="Tiempo (s)")
    plt.xlabel("Tamaño de entrada (n)")
    plt.ylabel("Tiempo de ejecución")
    plt.title("Complejidad temporal de merge_sort")
    plt.legend()
    plt.grid(True)
    plt.show()

    # Graficar memoria
    plt.figure(figsize=(10,5))
    plt.plot(tamanos, memorias, marker='o', color="red", label="Memoria (KB)")
    plt.xlabel("Tamaño de entrada (n)")
    plt.ylabel("Memoria usada (KB)")
    plt.title("Complejidad espacial de merge_sort")
    plt.legend()
    plt.grid(True)
    plt.show()

monedas = [1, 2, 5, 10, 20, 50]

subtotal = int(input("Ingresa el total a pagar: "))
dinero = int(input("Ingresa el dinero para pagar tu total: "))

print("="*10, "Cambio de monedas", "="*10)
print(f"Denominaciones disponibles: {monedas}")
print(f"Denominaciones ordenadas: {merge_sort(monedas)}")
monto = dinero - subtotal
print(f"Monto a devolver: {monto}")
cambio = cambio_recursivo(monto, merge_sort(monedas), 0)
print(f"Cambio entregado: {cambio}")
print(f"Cantidad de monedas usadas: {len(cambio)}")

# Ejecutar análisis
analizar_complejidad()
```

### *Función del cambio recursivo*
- Se ingresará el monto, las monedas disponibles y se definirá un indice para nuestra función
- Comprobará si el monto dado es 0, de ser así imprimirá la cadena.
- En caso de que nuestro índice sea menor o igual a la cantidad de elementos que conforman a nuestra lista se regresará la cadena.
- La moneda que evaluaremos será aquella que nos indique nuestro índice
#### Usos de la recursividad 
- En caso de que nuestro monto sea mayor a nuestra moneda actual se regresará lo siguiente: La moneda actual junto a la misma función con el monto, aal que se le restará la moneda que se evaluó.
- En caso contrario se llamará a la misma función, cuyo índice señalará al siguiente valor.

### *Función merge_sort*
- En caso de solo haber un solo elemento en la lista se regresará tal cual como se ingresó.
- Se determinará la posición media de la lista para posteriormente dividirla en dos cadenas diferentes.
- Se regresarán las dos cadenas que se generaron.
### *Función merge*
- Se evaluará cada elemento de ambas cadenas.
- Con todos los elementos ya ordenados, será unidos dentro de una sola cadena.
- Se regresará esta nueva cadena.


### *Programa principal*
- Se definirán las denominaciones disponibles en una caadena.
- Se ordenará nuestra cadena para que sus elementos estén en orden ascendente
- Se definirá el monto de dinero a devolver.
- Nuestro cambio se definirá con la función recursiva, ingresándo monto,  nuestra lista ordenada y un índice como sus parámetros
- Se imprimirá nuestro cambio junto a las monedas que fueron devueltas. 
- Se mostrarán las gráficas de complejidad del programa.
